###Limpieza de datos###

```python
import pandas as pd

df = pd.read_csv("../data/tipos_emergencia.csv")
print(df.shape)
df
```

```python
print(df.dtypes)
print(df["prioridad_base"].value_counts())
print(df["requiere_medico"].value_counts())
print(df["categoria"].value_counts())
```

Tocamos ahora columna "codigo"

```python
df["codigo"].str.strip().str.lower().value_counts().head(10)
```

Ahora vamos a por "prioridad base" y a por "requiere medico" 

```python
df_clean = df.copy()

df_clean["prioridad_base"] = df_clean["prioridad_base"].str.strip().str.lower()

mapa_medico = {"1": True, "true": True, "si": True, "yes": True,
               "0": False, "false": False, "no": False}
df_clean["requiere_medico"] = (
    df_clean["requiere_medico"].str.strip().str.lower().map(mapa_medico)
)

print(df_clean["prioridad_base"].value_counts())
print(df_clean["requiere_medico"].value_counts(dropna=False))
```
